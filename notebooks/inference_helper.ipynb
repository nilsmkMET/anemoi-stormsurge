```python
# import argparse
import sys

def replace_text_in_file(input_file, output_file, old_text, new_text):
    """
    Reads a file, replaces all occurrences of a specific string, and writes
    the content to a new or the same file.

    Args:
        input_file (str): The path to the input text file.
        output_file (str): The path to the output text file.
        old_text (str): The string to be replaced (e.g., 'YYYY').
        new_text (str): The string to replace it with (e.g., '2024').
    """
    try:
        with open(input_file, 'r', encoding='utf-8') as f_in:
            content = f_in.read()

        new_content = content.replace(old_text, new_text)

        with open(output_file, 'w', encoding='utf-8') as f_out:
            f_out.write(new_content)
        
        print(f"Successfully replaced '{old_text}' with '{new_text}' in '{input_file}' and saved to '{output_file}'.")

    except FileNotFoundError:
        print(f"Error: The file '{input_file}' was not found.", file=sys.stderr)
    except Exception as e:
        print(f"An error occurred: {e}", file=sys.stderr)
```

```python
for date_val in date_range[:]:
    # Add one day to date_val
    if add_one_day:
        date_val += pd.Timedelta(days=1)
    print(date_val.strftime('%Y-%m-%d'))
    for step in steps:
        replace_text_in_file(os.path.join(idir, 'main-anemoi-infer-MAL.yaml'), os.path.join(odir, f'main-anemoi-infer-{date_val.strftime("%Y-%m-%d")}-{step}-{expname}.yaml'), 'DATEHERE', str(date_val.strftime('%Y-%m-%d')))
        replace_text_in_file(os.path.join(odir, f'main-anemoi-infer-{date_val.strftime("%Y-%m-%d")}-{step}-{expname}.yaml'), os.path.join(odir, f'main-anemoi-infer-{date_val.strftime("%Y-%m-%d")}-{step}-{expname}.yaml'), 'STEPHERE', f'{step:06d}')
        replace_text_in_file(os.path.join(odir, f'main-anemoi-infer-{date_val.strftime("%Y-%m-%d")}-{step}-{expname}.yaml'), os.path.join(odir, f'main-anemoi-infer-{date_val.strftime("%Y-%m-%d")}-{step}-{expname}.yaml'), 'EXPNAME', expname)
```
